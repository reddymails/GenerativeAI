```python
# Creates an OpenAI chat model
# Registers a custom math tool (add)
# Builds a LangChain agent with that tool
# Asks the agent a question
# The agent decides to call the tool
# Prints the final answer
#  In short: An AI agent that can reason and call tools

import os
from langchain_openai import ChatOpenAI
from langchain_core.tools import tool
from langchain.agents import create_agent
from dotenv import load_dotenv
# Load env
load_dotenv(r"C:\Rama\Learn\AI\.env")
if not os.getenv("OPENAI_API_KEY"):
    raise RuntimeError("OPENAI_API_KEY not loaded")


@tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

model = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)

agent = create_agent(
    model=model,
    tools=[add]
)

result = agent.invoke(
    {"messages": [("user", "What is 3 + 5?")]}
)

print(result["messages"][-1].content)
```
